RAG (Retrieval Augmented Generation)

LLM have training cut off date.
Also LLMs' have no access to private data.
Hallucinations while prompting a raw LLM.


Traditional LLM :

User question -> LLM -> Answer (limited knowledge)

RAG System :

User question -> Retrieve relevant docs -> (LLM + those retrieved docs) -> Answer (Accurate)

INDEXING:
(one-time setup)
Documents ->parsing & processing -> chunks -> embeddings -> vector DB
(Gather docs -> clean ->extract)

RETRIEVAL (PER QUERY)
question -> embedding -> search -> relevant chunks

AUGMENT:
prompt + relevant chunks + query

GENERATION:
Question -> context -> Generative LLM -> answer

An LLM has a limited amount of text/tokens it can process in one request.10,000 PDF documents,You cannot put all those documents into the prompt.

Context window is limited → you can't stuff your entire knowledge base into the prompt → RAG retrieves only the relevant information.

1)RAG: best of both, required vector DB setup.External knowledge , Retrieve relevant information from a database and put it into the prompt. Gives LLM access to private/current knowledge without retraining
2) FINE TUNING (internal of LLM model) :You can fine-tune a model so it learns the desired behavior/style/task, deep customization. But requires ML expertise, slow and expensive.Model parameters,weights changed, train the model further on examples, Deep customization of behavior/task/style.
3)PROMPT ENGINEERING:Gives the existing LLM better instructions/context, free fast, limited by context window,model behavior isn't fundamentally changed.

Vectors that are closer to each other represent similar concepts, they have similar semantic meaning.

Distance=Similarity
