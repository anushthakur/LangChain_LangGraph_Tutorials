LLM's have context limit, must fit within token window.

Smaller chunks precise retrieval, better relevance matching.

If chunks are too small, then they lose context (incomplete information) which is not good for the downstream processing.

If chunks are too large, then noise in retrieval(irrelevant content included)

Different Strategies for Chunking:

RecursiveCharacterTextSplitter

It doesn't immediately chop your text every 1000 characters.Instead, it tries to preserve the natural structure of the text.
RecursiveCharacterTextSplitter is more flexible because it can progressively try different separators.

CharacterTextSplitter

This is a simpler splitter that splits text based on a single specified separator.

TokenTextSplitter

This splits text based on tokens, rather than characters.
This can be useful when you specifically care about token limits

MarkdownHeaderTextSplitter

This is useful when your source is a Markdown file.
Splits text based on specified Markdown headers (e.g., #, ##).

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200, #Keep approximately 200 characters from the previous chunk in the next chunk.The repeated portion helps prevent important information from being lost at chunk boundaries.If there were no overlap, the relationship between the two pieces could be weaker.
    separators=["\n\n", "\n", ""] #TThe order matters.his tells the splitter where it should try to break the text.
) 
#So the splitter first tries "\n\n" to split at paragraph boundaries, then "\n" at a normal line break, then "" split at the character level.
#chunk_size here is generally measured using the splitter's length function, which by default counts characters, not tokens.

#Create chunks of roughly 1000 characters, keep about 200 characters of overlap between neighboring chunks, and preferably split at paragraph boundaries, then line boundaries, and finally individual characters if necessary.
```

An embedding model converts the sentence into a list of numbers called an embedding/vector.The numbers represent the semantic meaning of the text.

The number of tokens does not determine the number of numbers in the embedding vector.The number of numbers is determined by the embedding model's output dimension, not by the number of tokens.

Dimension:
Number of values in the resulting vector

Tokens tell us how the text is broken into pieces; embedding dimensions tell us how many numbers represent the resulting meaning.

PURPOSE: Represent meaning numerically so that we can search for similar content.

An inference model (often called a generative/LLM model in this context) takes information and generates an answer.

PURPOSE: Understand the input and generate a response.

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

def document_splitter():
    
    loader=PyPDFLoader("./docs/langchain_demo.pdf") 
    docs=loader.load()
    print(f"Loaded {len (docs)} documents from PDF.")

    #PyPDFLoader.load() usually creates one LangChain Document per PDF page.
    # LangChain's loader represents each page as a separate Document object.
    
    # A loader decides how it converts the source into LangChain Document objects. Different loaders can use different granularities.

    #  Loader        	   Common behavior
    # PyPDFLoader	      Often one Document per PDF page
    # TextLoader	      Typically one Document for the text file
    # WebBaseLoader	  Typically one Document per webpage/URL
    # DirectoryLoader	  Loads multiple files; each file can produce one or more Documents depending on the loader used
    # CSVLoader	      Commonly one Document per CSV row

    splitter=RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=50,
    )

    #SPLIT THE DOCS INTO CHUNKS
    split_docs=splitter.split_documents(docs)

    print(f"Split into {len(split_docs)} chunks")
    print(f"\nFirst chunk metadata: {split_docs[0].metadata}")
    print(f"First chunk content: {split_docs [0].page_content[:200]}...")
    print(f"\nLast chunk metadata: {split_docs [-1].metadata}")

if __name__=="__main__":
    document_splitter()
```
