Create a complete vector store setup chat

1. Take a list of text strings

2.Split them into chunks

3.Store them into Chroma

4.Return a configured Retriever

```python

from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document


def create_retriever(
    texts:list[str],
    chunk_size:int=500,
    chunk_overlap:int=50,
    k:int=3
):
    #create documents
    documents=[Document(page_content=t) for t in texts]

    #split
    splitter=RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
    )
    split_docs=splitter.split_documents(documents)

    #create vectorstore, (in-memory) for this excercise(In-memory vector store (temporary RAM),Persistent vector database (saved on disk/cloud))
    vectorstore=Chroma.from_documents(
        documents=split_docs,
        embedding=OpenAIEmbeddings(model="text-embedding-3-small")
    )
    #Return retriever
    return vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k":k},
    )


def excercise_vectorstore_setup():
    #TEST THE FUNCTION

    #Text
   SAMPLE_TEXTS = [
    """
    Python is a versatile programming language used in web development,
    data science, machine learning, and automation.
    It has a simple syntax that makes it easy to learn and read.
    """,

    """
    JavaScript is the language of the web.
    It runs in browsers and servers with Node.js.
    """,

    """
    Rust is a systems programming language focused on safety and performance.
    """
 ]

   retriever= create_retriever(SAMPLE_TEXTS,chunk_size=100,chunk_overlap=20,k=2)

   queries=[
        "what is good for web development?",
        "What is the safest?"
    ]

   for query in queries:
        print(f"Query: {query}")
        results = retriever.invoke(query)
        for doc in results:
            print(f" - {doc.page_content[:60]}...")

if __name__=="__main__":
    excercise_vectorstore_setup()
```
